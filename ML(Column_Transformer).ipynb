```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/credit_scoring - credit_scoring.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
x = df.drop(columns=['Loan Amount'])
y = df['Loan Amount']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=50)
```

```python
x_train['Type of Loan'].value_counts()
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
oe = OrdinalEncoder(categories=[['Male','Female']])
x_train_gender = oe.fit_transform(x_train[['Gender']])

x_test_gender = oe.fit_transform(x_test[['Gender']])
```

```python
x_train_gender.shape
```

```python
ohe = OneHotEncoder(drop='first',sparse_output=False)

x_train_marital_education_emp = ohe.fit_transform(x_train[['Marital Status','Education Level','Employment Status']])

x_test_marital_education_emp = ohe.fit_transform(x_test[['Marital Status','Education Level','Employment Status']])
```

```python
x_train_marital_education_emp.shape
```

```python
oe = OrdinalEncoder(categories=[['Auto Loan','Home Loan','Personal Loan']])
x_train_loan = oe.fit_transform(x_train[['Type of Loan']])

x_test_loan = oe.fit_transform(x_test[['Type of Loan']])
```

```python
x_train_loan.shape
```

```python
x_train_num = x_train.select_dtypes(include='number').values
```

```python
x_test_num = x_test.select_dtypes(include='number').values
```

```python
x_train_transform = np.concatenate((x_train_gender,x_train_loan,x_train_marital_education_emp,x_train_num),axis=1)
```

```python
x_train_transform.shape
```

COLUMN TRANSFORMER

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[
    ('tnf1',OrdinalEncoder(categories=[['Male','Female'],['Auto Loan','Personal Loan','Home Loan']]),['Gender','Type of Loan']),
    ('tnf2',OneHotEncoder(drop='first',sparse_output=False),['Marital Status','Education Level','Employment Status'])
],remainder='passthrough')
```

```python
transformer.fit_transform(x_train).shape
```
